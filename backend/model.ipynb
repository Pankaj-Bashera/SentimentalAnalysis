# GO Emotions - Emotional Analyser

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("go_emotions_dataset.csv")
```

```python
print(df)
```

```python
print(df.shape)
print(df.columns)
```

```python
drop_cols = ['id','example_very_unclear']
df = df.drop(columns = drop_cols, errors="ignore")
```

```python
df = df.dropna(subset=["text"])
```

```python
print(df.shape)
```

```python
import re

def clean_text(text):
    text = text.lower()
    text = re.sub(r"http\S+", "", text)   
    text = re.sub(r"\s+", " ", text).strip()
    return text

df["text"] = df["text"].apply(clean_text)
```

```python
emotion_cols = df.columns.tolist()
emotion_cols.remove("text")
```

```python
df = df[df[emotion_cols].sum(axis=1) > 0]
```

```python
df = df[df[emotion_cols].sum(axis=1) <= 3]
```

```python
label_counts = df[emotion_cols].sum().sort_values()
print(label_counts)
```

```python
texts = df["text"].values
labels = df[emotion_cols].values
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    texts, labels, test_size=0.2, random_state=42
)
```
